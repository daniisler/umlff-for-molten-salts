```python
import warnings

import numpy as np
from ase.io import read
from mace.calculators import MACECalculator

warnings.filterwarnings("ignore")

DATA = (
    "/Users/mariospoulis/Documents/Materials Science and Engineering 2026-2027/Internship/umlff-for-molten-salts/data"
)
calc_ss = MACECalculator(model_paths=DATA + "/SuperSalt-swa.model", device="cpu", default_dtype="float64")


test = read(DATA + "/Testing_2.extxyz", index=":")
de, fp, fr = [], [], []
for a in test:
    e_ref, f_ref = a.get_potential_energy(), a.arrays["force"]
    m = a.copy()
    m.calc = calc_ss
    de.append((m.get_potential_energy() - e_ref) / len(m))
    fp.append(m.get_forces().ravel())
    fr.append(f_ref.ravel())
print(f"SuperSalt on TEST_1 (ternary, {len(test)} frames):")
print(f"  energy MAE  = {np.mean(np.abs(de)) * 1000:.2f} meV/atom")
print(
    f"  force  RMSE = {np.sqrt(np.mean((np.concatenate(fp) - np.concatenate(fr)) ** 2)) * 1000:.2f} meV/Ang   (paper Test_1 = 19.6)"
)
```

```python
import warnings
from collections import Counter

from ase.io import read

warnings.filterwarnings("ignore")

DATA = (
    "/Users/mariospoulis/Documents/Materials Science and Engineering 2026-2027/Internship/umlff-for-molten-salts/data"
)

for f in ["Testing_2.extxyz", "Tessting_1.extxyz"]:
    frames = read(f"{DATA}/{f}", index=":")  # load all frames in each file to check the correct testing files
    ncat = [
        len(set(a.get_chemical_symbols()) - {"Cl"}) for a in frames
    ]  # count the cations in each frame, excluding Cl anions
    print(f"{f}:")
    print(f"   total frames          = {len(frames)}")
    print(f"   cations per structure = {dict(sorted(Counter(ncat).items()))}")
    print()
```
